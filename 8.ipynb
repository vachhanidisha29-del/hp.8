```python
import numpy as np

views = np.array([
    1000.0, 2500.0, np.nan, np.inf,
    5000.0, -np.inf, 7500.0
])


max_finite = np.max(views[np.isfinite(views)])

clean_views = np.nan_to_num(
    views,
    nan=0.0,
    posinf=max_finite,
    neginf=max_finite
)

clean_views
```
